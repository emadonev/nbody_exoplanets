```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('../input/triple_systems.csv')
data.head()
```

```python
planet_data = data.copy()
planet_data = planet_data.drop(['hostname'], axis=1)
planet_data = planet_data.drop(['sy_snum'], axis=1)
planet_data = planet_data.drop(['sy_pnum'], axis=1)
planet_data = planet_data.drop(['st_spectype'], axis=1)
planet_data = planet_data.drop(['st_teff'], axis=1)
planet_data = planet_data.drop(['st_rad'], axis=1)
planet_data = planet_data.drop(['st_mass'], axis=1)
planet_data = planet_data.drop(['sy_vmag'], axis=1)

planet_data.head()
```

```python
planet_data.to_csv('../output/planets_triples.csv')
```
